```python
# Task 1: Data Gathering & Combination

import sqlite3
import pandas as pd
import json
from IPython.display import display

# ==========================================
# 1. Connect to the Library Database
# ==========================================

connection = sqlite3.connect("library.db")

# Load the three database tables
members = pd.read_sql_query(
    "SELECT * FROM members",
    connection
)

books = pd.read_sql_query(
    "SELECT * FROM books",
    connection
)

checkouts = pd.read_sql_query(
    "SELECT * FROM checkouts",
    connection
)

print("Database data loaded successfully.")
print("Members:", len(members))
print("Books:", len(books))
print("Checkouts:", len(checkouts))


# ==========================================
# 2. Read Books JSON File
# ==========================================

with open("books.json", "r") as file:
    book_data = json.load(file)

book_details = pd.DataFrame(book_data)

print("\nBooks JSON loaded successfully.")
print("JSON rows:", len(book_details))


# ==========================================
# 3. Question 1: Checkouts for Each Member
# ==========================================

q1 = pd.read_sql_query("""
SELECT
    m.member_id,
    m.first_name,
    m.last_name,
    COUNT(c.checkout_id) AS checkout_count
FROM members m
LEFT JOIN checkouts c
    ON m.member_id = c.member_id
GROUP BY
    m.member_id,
    m.first_name,
    m.last_name
ORDER BY checkout_count DESC;
""", connection)

print("\nQuestion 1: Checkouts for Each Member")
display(q1)


# ==========================================
# 4. Question 2: Books by Authors Matching Darwish
# ==========================================

q2 = pd.read_sql_query("""
SELECT *
FROM books
WHERE author LIKE '%Darwish%';
""", connection)

print("\nQuestion 2: Books by Authors Matching Darwish")
display(q2)


# ==========================================
# 5. Question 3: Top 5 Most Popular Books
# ==========================================

q3 = pd.read_sql_query("""
SELECT
    b.book_id,
    b.title,
    b.author,
    COUNT(c.checkout_id) AS checkout_count
FROM books b
LEFT JOIN checkouts c
    ON b.book_id = c.book_id
GROUP BY
    b.book_id,
    b.title,
    b.author
ORDER BY checkout_count DESC
LIMIT 5;
""", connection)

print("\nQuestion 3: Top 5 Most Popular Books")
display(q3)


# ==========================================
# 6. Question 4: Top 10 Active Readers
# ==========================================

q4 = pd.read_sql_query("""
SELECT
    m.member_id,
    m.first_name,
    m.last_name,
    COUNT(c.checkout_id) AS checkout_count
FROM members m
JOIN checkouts c
    ON m.member_id = c.member_id
GROUP BY
    m.member_id,
    m.first_name,
    m.last_name
ORDER BY checkout_count DESC
LIMIT 10;
""", connection)

print("\nQuestion 4: Top 10 Active Readers")
display(q4)


# ==========================================
# 7. Question 5: Maadi Activity
#    Excluding the 10 Newest Checkouts
# ==========================================

q5 = pd.read_sql_query("""
SELECT
    m.member_id,
    m.first_name,
    m.last_name,
    m.neighborhood,
    c.checkout_id,
    c.book_id,
    c.checkout_date,
    c.return_date
FROM members m
JOIN checkouts c
    ON m.member_id = c.member_id
WHERE m.neighborhood = 'Maadi'
AND c.checkout_id NOT IN (
    SELECT checkout_id
    FROM checkouts
    ORDER BY checkout_date DESC
    LIMIT 10
)
ORDER BY c.checkout_date DESC;
""", connection)

print("\nQuestion 5: Maadi Activity Excluding 10 Newest Checkouts")
display(q5)


# ==========================================
# 8. Combine Members + Checkouts
# ==========================================

stage_1 = pd.merge(
    members,
    checkouts,
    on="member_id",
    how="inner"
)

print("\nStage 1 rows:", len(stage_1))


# ==========================================
# 9. Add Book Details from JSON
# ==========================================

stage_2 = pd.merge(
    stage_1,
    book_details,
    on="book_id",
    how="left"
)

print("Stage 2 rows:", len(stage_2))


# ==========================================
# 10. Read Summer Checkout HTML
# ==========================================

# ==========================================
# 10. Read Summer Checkout HTML
# ==========================================

summer = pd.read_html("summer_checkouts.html")[0]

print("Summer checkout columns:")
print(summer.columns.tolist())

print("Summer checkout rows:", len(summer))


# ==========================================
# 11. Rename Summer Columns
# ==========================================

summer = summer.rename(columns={
    "Member ID": "member_id",
    "Book ID": "book_id",
    "Checkout Date": "checkout_date"
})


# ==========================================
# 12. Add Missing Columns
# ==========================================

summer["first_name"] = pd.NA
summer["last_name"] = pd.NA
summer["neighborhood"] = pd.NA
summer["checkout_id"] = pd.NA
summer["return_date"] = pd.NA
summer["genre"] = pd.NA
summer["publication_year"] = pd.NA


# Put the columns in the same order
common_columns = [
    "member_id",
    "first_name",
    "last_name",
    "neighborhood",
    "checkout_id",
    "book_id",
    "checkout_date",
    "return_date",
    "genre",
    "publication_year"
]

summer_final = summer[common_columns]


# ==========================================
# 13. Combine Main Data + Summer Data
# ==========================================

stage_2_final = stage_2[common_columns]

final_data = pd.concat(
    [stage_2_final, summer_final],
    ignore_index=True
)

# Remove exact duplicates
final_data = final_data.drop_duplicates()


# ==========================================
# 14. Save Task 1 Dataset
# ==========================================

final_data.to_csv(
    "task1_combined_data.csv",
    index=False
)


# ==========================================
# 15. Final Task 1 Results
# ==========================================

print("\nTask 1 is complete!")
print("Final rows:", len(final_data))
print("Final columns:", final_data.columns.tolist())

display(final_data.head())
# Create Task 1 SQL Answers file

sql_answers = """
Task 1: SQL Answers

Question 1: Checkouts for Each Member
The query counts the number of checkouts for every library member, including members with zero checkouts.

Question 2: Books by Authors Matching Darwish
The query searches for books where the author's name contains "Darwish".
The results were:
- Poetry by Darwish — Mahmoud Darwish
- The Silver Kite — Amina Darwish
- Desert Compass — Amina Darwish

Question 3: Top 5 Most Popular Books
The five most popular books by checkout count were:
1. The Silver Kite — 57 checkouts
2. Fossils and Fireflies — 55 checkouts
3. Circuits for Beginners — 46 checkouts
4. Kites Over Cairo — 38 checkouts
5. Storms and Sailboats — 25 checkouts

Question 4: Top 10 Active Readers
The query identifies the ten members with the highest number of checkouts.
The most active reader was Aya Wahba with 25 checkouts, followed by Sherif Saleh with 21 checkouts.

Question 5: Maadi Activity Excluding the 10 Newest Checkouts
The query shows checkout activity for members from Maadi while excluding the ten newest checkouts.
The query returned 104 records.
"""

with open("task1_sql_answers.txt", "w") as file:
    file.write(sql_answers)

print("Created: task1_sql_answers.txt")
```

```python
# Task 2: Data Integrity

import pandas as pd
import sqlite3

# Load Task 1 combined data
data = pd.read_csv("task1_combined_data.csv")

print("Task 1 rows:", len(data))

# ==========================================
# 1. Missing Values
# ==========================================

print("\n1. Missing Values")

missing = data.isnull().sum()
print(missing[missing > 0])


# ==========================================
# 2. Exact Duplicates
# ==========================================

print("\n2. Exact Duplicates")

duplicates = data.duplicated().sum()
print("Number of exact duplicates:", duplicates)

# Remove exact duplicates
data = data.drop_duplicates()


# ==========================================
# 3. Standardize Text Values
# ==========================================

print("\n3. Text Values")

for column in data.select_dtypes(include="object").columns:
    data[column] = data[column].str.strip()

# Standardize neighborhood names
if "neighborhood" in data.columns:
    data["neighborhood"] = data["neighborhood"].str.title()

# Fill missing text information
text_columns = [
    "first_name",
    "last_name",
    "neighborhood",
    "genre",
    "publication_year"
]

for column in text_columns:
    if column in data.columns:
        data[column] = data[column].fillna("Not provided")

print("Text values have been standardized and missing text values have been handled.")


# ==========================================
# 4. Check Unmatched Member IDs
# ==========================================

print("\n4. Unmatched Member IDs")

connection = sqlite3.connect("library.db")

members = pd.read_sql_query(
    "SELECT member_id FROM members",
    connection
)

valid_member_ids = set(members["member_id"])

unmatched = data[
    ~data["member_id"].isin(valid_member_ids)
    & data["member_id"].notna()
]

print("Unmatched records:", len(unmatched))

# Remove unmatched records
data = data[
    data["member_id"].isin(valid_member_ids)
    | data["member_id"].isna()
]


# ==========================================
# 5. Final Checks
# ==========================================

print("\nFinal missing values:")
print(data.isnull().sum())

print("\nFinal exact duplicates:", data.duplicated().sum())

print("\nNeighborhood values:")
print(data["neighborhood"].value_counts())


# ==========================================
# 6. Save Cleaned Dataset
# ==========================================

data.to_csv(
    "task2_cleaned_data.csv",
    index=False
)

print("\nTask 2 is complete!")
print("Final rows:", len(data))
print("File created: task2_cleaned_data.csv")
```

```python
# Task 3: Data Fairness Analysis

import pandas as pd

# Load the cleaned dataset
data = pd.read_csv("task2_cleaned_data.csv")

# Calculate total members and total checkouts for each neighborhood
fairness_analysis = data.groupby("neighborhood").agg(
    total_members=("member_id", "nunique"),
    total_checkouts=("checkout_id", "count")
).reset_index()

# Display the results
print("Neighborhood Fairness Analysis:")
display(fairness_analysis)
```
